```python
# ==========================================
# Import Required Libraries
# ==========================================

import joblib
import pandas as pd
```

```python
# ==========================================
# Load Trained Random Forest Model
# ==========================================

model = joblib.load("pcos_model.pkl")

print("Model Loaded Successfully!")
```

```python
# ==========================================
# Create Sample Patient Data
# ==========================================

sample_data = pd.DataFrame({

    'Age (yrs)': [24],
    'Weight (Kg)': [72],
    'Height(Cm)': [160],
    'BMI': [28.1],
    'Blood Group': [15],
    'Cycle(R/I)': [4],
    'Cycle length(days)': [45],
    'Weight gain(Y/N)': [1],
    'hair growth(Y/N)': [1],
    'Skin darkening (Y/N)': [1],
    'Hair loss(Y/N)': [1],
    'Pimples(Y/N)': [1],
    'Fast food (Y/N)': [1],
    'Reg.Exercise(Y/N)': [0]

})

sample_data
```

```python
# ==========================================
# Predict PCOS
# ==========================================

prediction = model.predict(sample_data)

print("Prediction :", prediction[0])
```

```python
# ==========================================
# Display Prediction Result
# ==========================================

if prediction[0] == 1:
    print("⚠️ High Risk of PCOS")
else:
    print("✅ Low Risk of PCOS")
```

```python
# ==========================================
# Prediction Probability
# ==========================================

probability = model.predict_proba(sample_data)

print(probability)
```

```python
# ==========================================
# Display Prediction Confidence
# ==========================================

print("Probability of No PCOS : {:.2f}%".format(probability[0][0]*100))
print("Probability of PCOS    : {:.2f}%".format(probability[0][1]*100))
```
